```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from PIL import Image 

digits = load_digits()
```

```python
X = digits.data
y = digits.target

X_train, X_test, y_train, y_test = train_test_split(
    X,y, test_size = 0.2, random_state = 234
)
```

```python
model = SVC(kernel = "rbf")

model.fit(X_train, y_train)
```

```python
# image_path = input("Enter path of handwritten digit image: ")
img = Image.open(r"C:\Users\athni\OneDrive\Desktop\SPPU-BE-AIDS-LabPracticals\CL-I\Machine Learning\Assignment 3\testing images\Seven.jpg").convert("L")
img = img.resize((8,8))

img_array = np.array(img)
img_array = 255 - img_array

img_array = (img_array / 255.0) * 16
img_array = img_array.reshape(1,-1)
prediction = model.predict(img_array)

plt.imshow(img_array.reshape(8,8),cmap = "gray")
plt.title("Predicted Digit: " + str(prediction[0]))
plt.axis("off")
plt.show()

print("Predicted Digit:", prediction[0])
```
