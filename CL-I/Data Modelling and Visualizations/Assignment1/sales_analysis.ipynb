```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

salesCSV = pd.read_csv("sales_data.csv")
salesJSON = pd.read_json("sales_data.json")
salesExcel = pd.read_excel("sales_data.xlsx")
```

```python
print("CSV Data:")
print(salesCSV.info())
print("\nJSON Data:")
print(salesJSON.info())    
print("\nExcel Data:")
print(salesExcel.info())
```

```python
#Data Cleaning and Preprocessing
print("\nMissing Values in CSV:")
print(salesCSV.isnull().sum())

print("\nMissing Values in Excel:")
print(salesExcel.isnull().sum())

print("\nMissing Values in JSON:")
print(salesJSON.isnull().sum())
```

```python
DataList = [salesCSV, salesJSON, salesExcel]
def cleaning_data(data):
    # Drop missing values
    data.dropna(inplace=True)
    if "date" in data.columns:
        data["date"] = pd.to_datetime(
            data["date"]
        )
    return data
for i in DataList:
    cleaning_data(i)
```

```python
salesCSV["source"] = "CSV"
salesExcel["source"] = "Excel"
salesJSON["source"] = "JSON"

df = pd.concat(
    [salesCSV, salesExcel, salesJSON],
    ignore_index=True
)

print(df.head())

print("\nUnified Dataset Shape:", df.shape)

print("\nUnified Dataset Columns:")
print(df.columns)
```

```python
if "Quantity" in df.columns and "Price" in df.columns:

    df["total_sales"] = (
        df["Quantity"] * df["Price"]
    )


# Create year and month columns from date
if "Date" in df.columns:

    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df["Year"] = df["Date"].dt.year
    df["Month"] = df["Date"].dt.month
    df["Month_name"] = df["Date"].dt.month_name()
print(df.head())
```

```python
# ------------------------------------------------------------
# 6. SALES ANALYSIS
# ------------------------------------------------------------

if "total_sales" in df.columns:

    # Total sales
    total_sales = df["total_sales"].sum()

    # Average order value
    average_order_value = df["total_sales"].mean()

    print("\n========== SALES METRICS ==========")

    print(
        "Total Sales:",
        round(total_sales, 2)
    )

    print(
        "Average Order Value:",
        round(average_order_value, 2)
    )
```
