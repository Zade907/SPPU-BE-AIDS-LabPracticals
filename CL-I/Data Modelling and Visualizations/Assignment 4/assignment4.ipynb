```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import LabelEncoder
df = pd.read_csv("realestate_prices.csv")
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r'[^a-z0-9]+', '_', regex=True)
    .str.strip('_')
)
print(df.head())
print(df.info())
```

```python
df.columns
```

```python
from streamlit import columns


df.isnull().sum()
# #if null values
# df.ffill()
# df.bfill()
for col in df.columns:
    if df[col].dtype == 'object' or df[col].dtype == 'str':
        df[col] = df[col].fillna(df[col].mode().iloc[0]) if not df[col].mode().empty else df[col]
    else:
        df[col] = df[col].fillna(df[col].median())
#df.dropna()
```

```python
# Merge df2 with df on the 'id' column
df2 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'new_column': ['A', 'B', 'C', 'D']
})
df2.merge(df, how = 'left', on = 'id')
```

```python
df = df[df['sqft_above'] > 750]
df
```

```python
le = LabelEncoder()
df['waterfront'] = le.fit_transform(df['waterfront'])
```

```python
if 'price' in df.columns:
    if 'yr_built' in df.columns:
        print("\nAverage Sale Price by Year Built:")
        print(df.groupby('yr_built')['price'].mean())
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print("\nDataset Shape After Outlier Removal:", df.shape)

# Save cleaned dataset
df.to_csv("Cleaned_RealEstate_Prices.csv", index=False)

print("\nData wrangling completed successfully.")
```
